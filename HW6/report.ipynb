### Problem 1 
Perform a DFS on a graph above, starting at a vertex not used in the lecture example. Choose a different edge exploration order so that the traversal order is different from the one in the lecture notes.

In the lecture example, the starting vertex is chosen at `a`. For my DFS traversal of the graph above, I choose to start with vertex `h`.          

Starting DFS at `h`, we can get `h -> i -> f -> g`. Because at `g`, the only outgoing edge is `g -> i`, but `i` has already been discovered, so DFS backtracks.             

Since the graph is directed and not every vertex is reachable from h, we continue the DFS with the remaining unvisited vertices, using alphabetical order for the next unvisited vertex, which is vertex `a`. For vertex `a`, it has `a -> d` and `a -> b`.         

We start with `a -> d`. We can get `a -> d -> e`. Similarly, at `e`, both outgoing edges go to already discovered vertices, so DFS backtracks to `a`. Next, we continue from `a -> b`. We can get `a -> b -> c`. Similarly, at `c`, the edge `c -> a` goes to an already discovered vertex, so DFS finishes.        

Therefore, since the DFS finished, we can get the resulting DFS discovery order was `h, i, f, g, a, d, e, b, c`.
